##Coloring Function

```python
colors = ["Red", "Green", "Blue", "Yellow"]

districts = list(graph.keys())

color_assignment = {}
```

##create a function

```python
def is_safe(district, color):

    for neighbour in graph[district]:

        if color_assignment.get(neighbour) == color:
            return False

    return True
```

##Backtracking Function

```python
def map_coloring(index=0):

    if index == len(districts):
        return True

    district = districts[index]

    for color in colors:

        if is_safe(district, color):

            color_assignment[district] = color

            if map_coloring(index + 1):
                return True

            del color_assignment[district]

    return False
```

##Run the algorithm

```python
if map_coloring():
    print("Map coloring successful!")
else:
    print("No valid coloring found.")
```

##Display the assignments

```python
for district, color in color_assignment.items():
    print(f"{district}: {color}")
```

##Visualization

```python
G = nx.Graph()

for district, neighbours in graph.items():

    for neighbour in neighbours:
        G.add_edge(district, neighbour)

color_map = {
    "Red": "red",
    "Green": "green",
    "Blue": "blue",
    "Yellow": "yellow"
}

node_colors = [
    color_map[color_assignment[node]]
    for node in G.nodes()
]

plt.figure(figsize=(16, 12))

pos = nx.spring_layout(
    G,
    seed=42
)

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color=node_colors,
    node_size=1500,
    font_size=7
)

plt.title("Telangana District Map Coloring using CSP")
plt.show()
```
